# Task: Loops and Testing

For each subtask, write a function that returns the string values given.

````python
import ipytest
ipytest.autoconfig()
````

## 1. `line()`

`line(1)` returns 
```
*
```

`line(2)` returns 
```
**
```

`line(5)` returns 
```
*****
```

````python
def line(n):
    return n * '*'
````

````python
%%ipytest

def test_line():
    assert line(1) == '*'
    assert line(2) == '**'
    assert line(5) == '*****'
````

````python
print(line(10))
````

## 2. `triangle()`

`triangle(1)` returns 
```
*
```

`triangle(2)` returns 
```
*
**
```

`triangle(3)` returns 
```
*
**
***
```


````python
def triangle(n):
    output = ''
    for i in range(1, n + 1):
        output = output + line(i) + '\n'
    return output[:-1]
````

````python
%%ipytest

def test_triangle():
    assert triangle(1) == '*'
    assert triangle(2) == '*\n**'
    assert triangle(3) == '*\n**\n***'
````

````python
print(triangle(10))
````

## 3. `square()`

`square(1)` returns 
```
*
```
`square(2)` returns
```
**
**
```
`square(3)` returns
```
***
* *
***
```
`square(4) returns
```
****
*  *
*  *
****
```

````python
def square(n):
    if(n == 1):
        return '*'
    output = line(n) + '\n'
    for i in range(n - 2):
        output = output + '*' + (n - 2) * ' ' + '*' + '\n'
    output = output + line(n)
    return output
````

````python
%%ipytest

def test_square():
    assert square(1) == '*'
    assert square(2) == '**\n**'
    assert square(3) == '***\n* *\n***'
    assert square(4) == '****\n*  *\n*  *\n****'
````

````python
print(square(10))
````

## 4. `mosaic()`

`mosaic(1)` returns
```
*
```
`mosaic(2)` returns
```
*.
.*
```
`mosaic(5)` returns
```
*.*.*
.*.*.
*.*.*
.*.*.
*.*.*
```

````python
def mosaic(n):
    output = ''
    for i in range(n):
        a = '*'
        b = '.'
        if(i % 2 == 1):
            a = '.'
            b = '*'
        for j in range(n):
            if(j % 2 == 0):
                output = output + a
            else:
                output = output + b
        output = output + '\n'
    return output[:-1]


````

````python
%%ipytest

def test_mosaic():
    assert mosaic(1) == '*'
    assert mosaic(2) == '*.\n.*'
    assert mosaic(5) == '*.*.*\n.*.*.\n*.*.*\n.*.*.\n*.*.*'
````

````python
print(mosaic(10))
````
