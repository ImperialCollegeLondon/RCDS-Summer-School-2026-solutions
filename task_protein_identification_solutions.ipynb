# Task: protein identification

Proteins isolated from a biological sample can be identified using techniques that separate them by their mass.

Given the *S. cerevisiae* proteome (set of all proteins encoded by the yeast genome) in `data/S_cerevisiae_proteins.fasta`, make a CSV output file with two columns:

```
protein_ID, protein_mass
```

You will need the data file `data/amino_acid_table.csv` which contains the molecular mass (in Daltons) for each amino acid.

We can break this task down into steps:

## 1.
Read in the data from `data/amino_acid_table.csv` and create a dictionary `aa_mass` that gives the molecular mass (value) for each amino acid letter (key).

````python
aa_mass = {}
with open('../course_content/data/amino_acid_table.csv','r') as f:
    f.readline()  # remove the table header
    for line in f:
        data = line.split(',')  # make a list of data items for this line
        aa_mass[data[2]] = float(data[4])
print(aa_mass)
````

## 2.
Write a function `protein_mass()` that takes a protein sequence and returns its total mass.

````python
def protein_mass(seq):
    tot =0
    for aa in seq:
        if(aa in aa_mass):  # check that this is a valid symbol
            tot = tot + aa_mass[aa]
    return tot
````

What is `protein_mass('TTVSDDFEGYVYTFDNNL')`?

````python
protein_mass('TTVSDDFEGYVYTFDNNL')
````

## 3.
Reading each protein sequence from the FASTA file, create the output file as described above.

````python
# from 'working with files' notebook

def read_multi_fasta(path):
    fasta_dict = {}
    with open(path, 'r') as f:
        descr = f.readline()[1:].rstrip()
        seq = ''
        for line in f:
            if(line.startswith('>')):
                fasta_dict[descr] = seq
                descr = line[1:].rstrip()
                seq = ''
            else:
                seq = seq + line.rstrip()
    return fasta_dict
````

````python
db = read_multi_fasta('../course_content/data/S_cerevisiae_proteins.fasta')
````

````python
with open('../course_content/output/protein_masses.csv','w') as f_out:
    f_out.write('"protein_ID","protein_mass"\n')
    for descr in db:
        pid = descr.split(" ")[0]
        f_out.write(pid + ',' + str(protein_mass(db[descr])) + '\n')
    
````
