All test tubes produced by your factory conform to the following design:

The tube is formed of glass of a thickness $t$ which is constant over the whole surface.

Above the hemisphere is a cylinder of height $h$ and outer radius $r$.

### 1.

A standard test tube has these measurements in mm: 

$t = 1.2$, 

total width, $w = 2r = 13$, and 

total height, $H = h + r = 100$.

What is the maximum volume of liquid that a standard test tube can hold (in ml, to 1dp)?

*Hint*: think about making general functions to calculate the volume of a hemisphere and of a cylinder. Then you can write a function that uses these to calculate the total volume of a test tube, given $t$, $w$ and $H$.

volume of a cylinder = $\pi r^2 h$

volume of a hemisphere = $\frac{2}{3} \pi r^3$ 

1ml = 1 cubic centimetre = 1000 cubic millimitres


```python
PI = 3.1415927

def cylinder_volume(r, h):
    return PI * (r ** 2) * h

def hemisphere_volume(r):
    return 2/3 * PI * (r ** 3)

def tube_capacity(t, w, H):
    r = w / 2
    h = H - r
    # subtract the glass thickness from the radius to give the capacity.
    return cylinder_volume(r - t, h) + hemisphere_volume(r - t)

print(round(tube_capacity(0.12, 1.3, 10.0), 1)) # measurements in cm
```

### 2.

What volume of glass (in cubic cm, to 1dp) is required to make one standard test tube?

*Hint*: think about how to calculate this using the functions you built in part 1.


```python
# the volume of glass is found by subtracting the tube's capacity 
# from the volume enclosed by the outer surface of the glass

def glass_volume(t, w, H):
    r = w / 2
    h = H - r
    outer_volume = cylinder_volume(r, h) + hemisphere_volume(r)
    inner_volume = cylinder_volume(r - t, h) + hemisphere_volume(r - t)
    return outer_volume - inner_volume

print(round(glass_volume(0.12, 1.3, 10.0), 1)) # measurements in cm
```

### 3.

An important customer has just placed the following order for test tubes:

| tube type | $t$ | $w$ | $H$ | quantity |
| --- | --- | --- | --- | --- |
| mini | 1.0 | 10 | 75 | 5000 |
| standard | 1.2 | 13 | 100 | 10000 |
| giant | 1.5 | 20 | 150 | 1000 |

(all measurements in mm)

Allowing a 10% margin for wastage, how much glass (to the nearest kg) will you need to buy to fulfil this order?

The density of borosilicate glass is 2.23 g per cubic centimetre.

```python
GLASS_DENSITY = 2.23  # g/cm3
MARGIN = 0.1  # for wastage

glass_volume_mini = glass_volume(0.1, 1.0, 7.5)
glass_volume_standard = glass_volume(0.12, 1.3, 10.0)
glass_volume_giant = glass_volume(0.15, 2.0, 15.0)

total_glass_volume = 5000 * glass_volume_mini \
                   + 10000 * glass_volume_standard \
                   + 1000 * glass_volume_giant

total_glass_mass =  GLASS_DENSITY * total_glass_volume / 1000

total_glass_ordered = (1 + MARGIN) * total_glass_mass

print(round(total_glass_ordered))
```
